**Aplicação de distribuição binomial**

Suponha que desejamos calcular a probabilidade de x = 8 procedimentos de angioplastia bem-sucedidos em n = 10 pacientes com estenose unilateral da artéria renal. A probabilidade de uma angioplastia ser bem sucedida de cada vez é de 70%. Podemos calcular a probabilidade de uma angioplastia ser bem sucedida.
E 7, 8 ou 9?



```python
#@title **Vamos importar a biblioteca SciPy, que oferece uma interface robusta para distribuições.**

from scipy.stats import binom
```

```python
#@title **Para a probabilidade exata - Use binom.pmf (Probability Mass Function)**
n = 10  # Número de tentativas
p = 0.7  # Probabilidade de sucesso
k = 8  # Número de sucessos desejado

prob = binom.pmf(k, n, p)
print(f"Probabilidade de exatamente {k} angioplastias bem sucedidas {prob}")
```

```python
 #@title **Probabilidade acumulada - P(X≤k) - Use binom.cdf (Cumulative Distribution Function)**

n = 10
p = 0.7
k = 8

cum_prob = binom.cdf(k, n, p)
print(f"Probabilidade de no máximo {k} angioplastias bem sucedidas: {cum_prob}")
```

```python
 #@title **Probabilidade acumulada - P(X>k) - Calcule como o complemento de P(X≤k)**

n = 10
p = 0.5
k = 8

cum_prob = 1 - binom.cdf(k, n, p)
print(f"Probabilidade de no mínimo {k} angioplastias bem sucedidas: {cum_prob}")
```

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parâmetros da distribuição binomial
n = 10
p = 0.7
x = range(n + 1)  # Valores possíveis de X
probs = binom.pmf(x, n, p)

# Criar o gráfico de barras
plt.bar(x, probs, color='blue', alpha=0.7)

# Adicionar os valores das probabilidades no topo de cada barra
for i, prob in enumerate(probs):
    plt.text(i, prob + 0.01, f'{prob:.2f}', ha='center', fontsize=9, color='black')

# Configuração dos rótulos e título
plt.xlabel('Número de Sucessos')
plt.ylabel('Probabilidade')
plt.title('Distribuição Binomial')

# Exibir o gráfico
plt.show()
```

```python
#@title **Este código retorna o menor valor k tal que a probabilidade cumulativa da distribuição binomial P(X≤k) seja maior ou igual a 0.95, com 10 tentativas (e probabilidade de sucesso de 0.7)**


n = 10  # Número de tentativas
p = 0.7  # Probabilidade de sucesso
q = 0.95  # Nível de probabilidade cumulativa

quantile = binom.ppf(q, n, p)
print(f"Quantil: {quantile}")

```

```python
#@title **A biblioteca SciPy oferece ferramentas completas para manipular a distribuição de Poisson. Importe o módulo com:**

from scipy.stats import poisson
```

Suponha que a incidência de um certo tipo de câncer em uma determinada região é de 250 casos por ano. Qual é a probabilidade de haver exatamente 135 casos de câncer nos próximos 6 meses?


```python
#@title **Probabilidade exata P(X=k) - Use a Probability Mass Function (PMF) com poisson.pmf.**
mu = 125  # Média (lambda)
k = 135  # Número de eventos desejados

prob = poisson.pmf(k, mu)
print(f"Probabilidade de ocorrência de exatamente {k} cânceres nos próximos 6 meses: {prob}")
```

```python
#@title **Probabilidade acumulada P(X≤k) Use a Cumulative Distribution Function (CDF) com poisson.cdf.**
mu = 125  # Média (lambda)
k = 135  # Número de eventos desejados

cum_prob = poisson.cdf(k, mu)
print(f"Probabilidade de ocorrência de até {k} eventos: {cum_prob}")
```

```python
#@title **Probabilidade acumulada P(X>k) Use a Cumulative Distribution Function (CDF) com poisson.cdf.**
mu = 125  # Média (lambda)
k = 135  # Número de eventos desejados

cum_prob = 1 - poisson.cdf(k, mu)
print(f"Probabilidade de ocorrência de no mínimo {k} eventos: {cum_prob}")
```

```python
#@title **Visualização da distribuição**
import matplotlib.pyplot as plt

mu = 125  # Média (lambda)
x = range(0, 200)  # Valores possíveis de X
probs = poisson.pmf(x, mu)

plt.bar(x, probs, color='blue', alpha=0.7)
plt.xlabel('Número de Eventos')
plt.ylabel('Probabilidade')
plt.title('Distribuição de Poisson')
plt.show()
```
